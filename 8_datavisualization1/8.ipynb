```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
data = pd.read_csv('https://raw.githubusercontent.com/dphi-official/Datasets/master/titanic_data.csv')
data
```

```python
data.shape
```

```python
data.describe()
```

```python
data.describe(include = 'object')
```

```python
data.isnull().sum()
```

```python
data['Age'] = data['Age'].fillna(np.mean(data['Age']))
data['Cabin'] = data['Cabin'].fillna(data['Cabin'].mode()[0])
data['Embarked'] = data['Embarked'].fillna(data['Embarked'].mode()[0])
data.isnull().sum()
```

```python
data['Survived'] = data['Survived'].astype(int)
```

```python
sns.countplot(x=data['Survived'])
plt.show()
```

```python
sns.countplot(x=data['Pclass'])
plt.show()
```

```python
sns.countplot(x=data['Embarked'])
plt.show()
```

```python
sns.countplot(x=data['Sex'])
plt.show()
```

```python
sns.boxplot(x=data['Age'])
plt.show()
```

```python
sns.catplot(x= 'Pclass', y = 'Age', data=data, kind = 'box')
```

```python
sns.catplot(
    x='Pclass',
    y='Fare',
    hue='Pclass',
    data=data,
    kind='strip',
    palette='deep'
)
plt.show()
```

```python
sns.catplot(
    x='Sex',
    y='Fare',
    hue='Sex',
    data=data,
    kind='strip',
    palette='deep'
)

plt.show()
```

```python
sns.catplot(x= 'Sex', y = 'Age',hue='Sex', data=data, kind = 'strip', palette='deep')
plt.show()
```

```python
numeric_columns = data.select_dtypes(include=[np.number]).columns  # Select numeric columns

# Filter the DataFrame to include only numeric columns
df_numeric = data[numeric_columns]

# Calculate correlation matrix
tc = df_numeric.corr()

sns.heatmap(tc, cmap="YlGnBu")
plt.title('Correlation')
```
