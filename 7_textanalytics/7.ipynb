```python
import pandas as pd 
import nltk 
import math 
from nltk.corpus import stopwords 
from nltk.stem import PorterStemmer, LancasterStemmer, WordNetLemmatizer 
from nltk.tokenize import word_tokenize, sent_tokenize 
```

```python
nltk.download("universal_tagset") 
nltk.download("punkt") 
nltk.download("stopwords") 
nltk.download('punkt_tab') 
nltk.download("wordnet") 
nltk.download("averaged_perceptron_tagger_eng")
```

```python
text = """Scarlett O'Hara was not beautiful, but men seldom realized it when caught by her charm as the Tarleton twins were. In her face were too sharply blended the delicate features of her mother, a Coast aristocrat of French descent, and the heavy ones of her florid Irish father. But it was an arresting face, pointed of chin, square of jaw. Her eyes were pale green without a touch of hazel, starred with bristly black lashes and slightly tilted at the ends. Above them, her thick black brows slanted upward, cutting a startling oblique line in her magnolia-white skin--that skin so prized by Southern women and so carefully guarded with bonnets, veils and mittens against hot Georgia suns.""" 
print(text) 
```

```python
tokenized_text = nltk.word_tokenize(text) 
print("Tokenized Text:", tokenized_text)
```

```python
tokenized_sent = nltk.sent_tokenize(text) 
print("Tokenized Sentences:", tokenized_sent) 
```

```python
words = [nltk.word_tokenize(i) for i in nltk.sent_tokenize(text)] 
pos_tag = [nltk.pos_tag(i, tagset="universal") for i in words] 
print("POS Tagging:", pos_tag) 
```

```python
to_be_removed = set(stopwords.words("english")) 
stopWordsRemovedList = [ word for word in tokenized_text if word.lower() not in to_be_removed ] 
print("Stop Words Removed:", stopWordsRemovedList)
```

```python
stemmerPort = PorterStemmer() 
portStemmedWords = [stemmerPort.stem(i) for i in tokenized_text] 
print("Porter Stemmed Words:", portStemmedWords) 
```

```python
stemmerLan = LancasterStemmer() 
lanStemmedWords = [stemmerLan.stem(i) for i in tokenized_text] 
print("Lancaster Stemmed Words:", lanStemmedWords) 
```

```python
lemmatizer = WordNetLemmatizer() 
lemmatizedWords = [lemmatizer.lemmatize(i) for i in tokenized_text] 
print("Lemmatized Words:", lemmatizedWords)
```

```python
sent_1 = text 
sent_2 = "Seated with Stuart and Brent Tarleton in the cool shade of the porch of Tara, her father’s plantation, that bright April afternoon of 1861, she made a pretty picture." 
sent_3 = "On either side of her, the twins lounged easily in their chairs, squinting at the sunlight through tall mint-garnished glasses as they laughed and talked." 
```

```python
tk_sent_1 = nltk.word_tokenize(sent_1) 
tk_sent_2 = nltk.word_tokenize(sent_2) 
tk_sent_3 = nltk.word_tokenize(sent_3) 
tk_sent_1 = [word for word in tk_sent_1 if word.lower() not in to_be_removed] 
tk_sent_2 = [word for word in tk_sent_2 if word.lower() not in to_be_removed] 
tk_sent_3 = [word for word in tk_sent_3 if word.lower() not in to_be_removed]  
```

```python
def computeTF(wordDict, doc):
    tfDict = {}
    corpusCount = len(doc)
    for word, count in wordDict.items():
        tfDict[word] = count / float(corpusCount)     
    return tfDict 
```

```python
wordDict_1 = dict.fromkeys(tk_sent_1, 0) 
for word in tk_sent_1:
    wordDict_1[word] += 1 
```

```python
wordDict_2 = dict.fromkeys(tk_sent_2, 0) 
for word in tk_sent_2:
    wordDict_2[word] += 1 
```

```python
wordDict_3 = dict.fromkeys(tk_sent_3, 0) 
for word in tk_sent_3:
    wordDict_3[word] += 1 
```

```python
tfFirst = computeTF(wordDict_1, tk_sent_1) 
tfSecond = computeTF(wordDict_2, tk_sent_2) 
tfThird = computeTF(wordDict_3, tk_sent_3) 
```

```python
def computeIDF(docList):
    idfDict = {}
    N = len(docList)
    # Collect all unique words from all documents     
    all_words = set()     
    for doc in docList:
        all_words.update(doc.keys())
    # Initialize idfDict with all words     
    idfDict = dict.fromkeys(all_words, 0)      
    # Compute document frequency for each word
    for doc in docList:
        for word in doc:
            if doc[word] > 0:
                idfDict[word] += 1
    # Compute IDF     
    for word in idfDict:
        idfDict[word] = math.log10(N / (float(idfDict[word]) + 1))
    return idfDict   
idfs = computeIDF([wordDict_1, wordDict_2, wordDict_3])
```

```python
def computeTFIDF(tfBow, idfs):
    tfidf = {}     
    for word, val in tfBow.items():         
        tfidf[word] = val * idfs.get(word, 0)     
    return tfidf   
idfFirst = computeTFIDF(tfFirst, idfs) 
idfSecond = computeTFIDF(tfSecond, idfs) 
idfThird = computeTFIDF(tfThird, idfs) 
```

```python
# Convert to DataFrame 
idf_df = pd.DataFrame([idfFirst, idfSecond, idfThird]) 
print(idf_df)
```
